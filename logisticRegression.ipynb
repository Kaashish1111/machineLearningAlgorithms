```python
import numpy as np

X = np.array([
    [1, 2],
    [2, 3],
    [3, 1],
    [4, 3],
    [5, 4],
    [6, 2]
])

y = np.array([0, 0, 0, 1, 1, 1])
```

```python
n=len(X)
# intialise weight and bias
w = np.zeros(X.shape[1])
b=0

#intialise hyperparmeter
lr=0.1
epoches=100
```

```python
for i in range(epoches):
  z = X @ w + b
  y_pred = 1 / (1 + np.exp(-z))

  error = y_pred - y
  dw = X.T @ error / n
  db = np.sum(error) / n

  loss = -np.mean(
    y * np.log(y_pred) + (1-y) * np.log(1-y_pred)
  )

  w=w-lr*dw
  b=b-lr*db
```

```python
print("updated weights are : ",w)
print("updated intercept is: ",b)
```
