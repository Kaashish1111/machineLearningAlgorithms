```python
import numpy as np
import pandas as pd
```

```python
# Dataset
X = np.array([1, 2, 3, 4])
y = np.array([2, 4, 6, 8])
```

```python

# Initialize parameters
w = 0.0
b = 0.0
```

```python
# Hyperparameters
lr = 0.01
epochs = 1000
n = len(X)
```

```python
for i in range(epochs):
  y_pred=w*X+b
  error = y_pred - y
  cost = np.sum(error**2)/2*n
  dw=-np.sum(error * X)/n
  db=np.sum(error)/n

  w=w-lr*dw
  b=b-lr*db
```

```python
print("Final weight:", w)
print("Final bias:", b)
```
