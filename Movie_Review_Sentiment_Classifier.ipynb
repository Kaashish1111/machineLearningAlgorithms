```python
import numpy as np

with open('reviews.txt', 'r') as f:
    raw_reviews = f.readlines()

with open('labels.txt', 'r') as f:
    raw_labels = f.readlines()

print(len(raw_reviews))
print(len(raw_labels))

print(raw_reviews[0])
print(raw_labels[0])
```

```python
positive_reviews=raw_labels.count('positive\n')
negative_reviews=raw_labels.count('negative\n')
print(positive_reviews)
print(negative_reviews)
```

```python
sentence = "I love this movie and I love acting"

words = sentence.split()

unique_words = set(words)

print(unique_words)
print(len(unique_words))
```

```python
# Convert each review into a set of words
tokens = map(lambda x: set(x.split()), raw_reviews)

# Create an empty vocabulary
vocab = set()

# Iterate through every review
for sent in tokens:

    # Iterate through every word in the review
    for word in sent:

        # Add the word to vocabulary
        if len(word) > 0:
            vocab.add(word)

# Print vocabulary size
print("Vocabulary size:", len(vocab))
```

```python
vocab = list(vocab)
word2index = {}
for i, word in enumerate(vocab):
    word2index[word] = i
    words = ['cat', 'dog', 'movie']

for i, word in enumerate(words):
    print(i, word)
print(word2index['movie'])
print(len(word2index))
```

```python
import numpy as np

word2vector = {}

vocab_size = len(vocab)

for i, word in enumerate(vocab):

    vector = np.zeros(vocab_size, dtype=int)

    vector[i] = 1

    word2vector[word] = vector
```
